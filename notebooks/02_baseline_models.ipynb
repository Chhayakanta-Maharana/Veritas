# Doomsday Sentinel: AI-Powered Deepfake Forensics
## Notebook 02: Conventional Machine Learning Baseline

Evaluates conventional computer vision baselines (Logistic Regression and Random Forest) using handcrafted forensic features (color channel moments, Laplacian texture variance, Sobel gradient energy, and 2D Fourier magnitude spectrum).

```python
import os
import sys
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

os.chdir('..') if os.path.basename(os.getcwd()) == 'notebooks' else None
sys.path.insert(0, os.getcwd())
from src.models.baseline import BaselinePipeline, build_dataset_features
```

### 1. Load Partitions and Extract Handcrafted Forensic Features

```python
train_df = pd.read_csv('data/manifests/train_manifest.csv')
val_df = pd.read_csv('data/manifests/val_manifest.csv')
test_df = pd.read_csv('data/manifests/test_manifest.csv')

print('Extracting features...')
X_train, y_train = build_dataset_features(train_df)
X_val, y_val = build_dataset_features(val_df)
X_test, y_test = build_dataset_features(test_df)
print(f'Feature Matrix Shape: {X_train.shape}')
```

### 2. Fit and Evaluate Logistic Regression Baseline

```python
lr = BaselinePipeline(model_type='logistic_regression')
lr.fit(X_train, y_train)
lr_val = lr.evaluate(X_val, y_val)
lr_test = lr.evaluate(X_test, y_test)
print('Logistic Regression Val Metrics:', lr_val)
print('Logistic Regression Test Metrics:', lr_test)
```

### 3. Fit and Evaluate Random Forest Baseline

```python
rf = BaselinePipeline(model_type='random_forest')
rf.fit(X_train, y_train)
rf_val = rf.evaluate(X_val, y_val)
rf_test = rf.evaluate(X_test, y_test)
print('Random Forest Val Metrics:', rf_val)
print('Random Forest Test Metrics:', rf_test)
```

### 4. Baseline Performance Comparison

```python
comparison_df = pd.DataFrame({
    'Logistic_Regression': lr_test,
    'Random_Forest': rf_test
}).T
comparison_df
```
